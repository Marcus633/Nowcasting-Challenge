### Plot MSRE for all predictions of each country

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 7))

for country, msre_values in ensemble_msre_dict2.items():
    predicted_months = list(range(1, len(msre_values) + 1))
    mean_value = sum(msre_values) / len(msre_values)
    label = f"{country} ({mean_value:.4f})"
    plt.plot(predicted_months, msre_values, marker='o', label=label)

plt.xlabel('Predicted Months')
plt.ylabel('MSRE')
plt.title(r'MSRE of $\bf{Ensemble 2}$ for Different Countries over Predicted Months')
plt.legend(title="Countries", bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

### Boxplot of the results per model

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Convert the msre_dicts into a DataFrame for easier plotting
import pandas as pd

# Flatten the dict of dicts to get a DataFrame where each row is a country/model combination
data = []
for model, msre_dict in msre_dicts.items():
    for country, msre_value in msre_dict.items():
        data.append({"model": model, "country": country, "msre": msre_value})

df = pd.DataFrame(data)

# Plot a boxplot
plt.figure(figsize=(12, 6))
sns.boxplot(x="model", y="msre", data=df)
plt.title("Distribution (Boxplot) of weighted MSRE Values Across Models")
plt.xlabel("Model")
plt.ylabel("MSRE Value")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

### Heatmap (Matrix) of the results per country for all models

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
from matplotlib.colors import BoundaryNorm, ListedColormap

#Define the custom discrete color map (green, blue, red)
colors = ["mediumspringgreen", "skyblue", "lightcoral"]  # Define the discrete colors
cmap_name = "discrete_msre_cmap"
cmap = ListedColormap(colors)

# Define the MSRE thresholds
boundaries = [0, 0.0001, 0.001, 1]  # Define the boundaries for the bins
norm = BoundaryNorm(boundaries, cmap.N)

# Convert the msre_dicts into a matrix format
msre_matrix = {model: [msre_dict.get(country, float('nan')) for country in msre_dicts["ensemble1"].keys()]
               for model, msre_dict in msre_dicts.items()}

# Create DataFrame
df_heatmap = pd.DataFrame(msre_matrix, index=msre_dicts["ensemble1"].keys())

# Plot heatmap with the custom discrete color map
plt.figure(figsize=(13, 6))
sns.heatmap(df_heatmap, annot=True, cmap=cmap, cbar_kws={'label': 'MSRE Value'},
            linewidths=0.5, linecolor='gray', norm=norm, annot_kws={"size": 10, "weight": "bold", "color": "black"})
plt.title("Weighted MSRE Values Across Countries and Models", fontsize=16)
plt.xlabel("Model", fontsize=12)
plt.ylabel("Country", fontsize=12)
plt.tight_layout()
plt.show()
```

### Scatterplot of the model results per country
